# Proyecto BI Mejorado

Este notebook incorpora:
- **Exploración de Hiperparámetros**: Uso de `GridSearchCV` para optimizar el modelo (en este ejemplo se usa un RandomForestClassifier).
- **WordCloud Basado en Pesos Vectoriales (tf-idf)**: Se utiliza `TfidfVectorizer` para obtener los pesos de las palabras y generar una nube de palabras en función de estos.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Librerías para exploración de hiperparámetros
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier

# Librerías para procesamiento de texto y WordCloud
from sklearn.feature_extraction.text import TfidfVectorizer
from wordcloud import WordCloud
```

## 1. Exploración de Hiperparámetros con GridSearchCV

```python
# Ejemplo: Utilizando RandomForestClassifier. Reemplaza con tu modelo y tus datos.

# Si ya cuentas con tus conjuntos de datos X_train e y_train, reemplaza la siguiente sección
from sklearn.datasets import load_iris
data = load_iris()
X_train = data.data
y_train = data.target

# Inicializamos el modelo base
modelo = RandomForestClassifier(random_state=42)

# Definición de la rejilla de hiperparámetros
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 4, 6]
}

# Configuramos el GridSearchCV con validación cruzada de 5 folds
grid_search = GridSearchCV(estimator=modelo,
                           param_grid=param_grid,
                           scoring='accuracy',
                           cv=5,
                           verbose=2,
                           n_jobs=-1)

grid_search.fit(X_train, y_train)

print("Mejores parámetros encontrados:", grid_search.best_params_)

# Modelo con los mejores parámetros
modelo_mejorado = grid_search.best_estimator_

# Aquí puedes continuar con la evaluación o guardado del modelo
print("Modelo optimizado:", modelo_mejorado)
```

## 2. Generación de WordCloud Basado en Pesos Vectoriales (tf-idf)

```python
# Ejemplo: Supongamos que tienes un corpus de documentos en una lista
corpus = [
    "Este es el primer documento.",
    "Este documento es el segundo documento.",
    "Y este es el tercer documento.",
    "¿Es este el primer documento?"
]

# Convertimos el corpus en una matriz tf-idf
vectorizer = TfidfVectorizer(max_features=1000)
tfidf_matrix = vectorizer.fit_transform(corpus)
feature_names = vectorizer.get_feature_names_out()

# Calculamos el peso promedio de cada palabra en el corpus
avg_tfidf = np.mean(tfidf_matrix.toarray(), axis=0)
pesos = dict(zip(feature_names, avg_tfidf))

# Generamos el WordCloud utilizando los pesos tf-idf
wordcloud = WordCloud(width=800, height=400, background_color='white')
wordcloud = wordcloud.generate_from_frequencies(pesos)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.show()
```
